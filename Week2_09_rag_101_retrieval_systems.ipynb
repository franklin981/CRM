```python
# YOUR CODE HERE
# Implement the intermediate exercise

pass  # Replace with your implementation
from typing import List, Dict, Any
import re
import html
from urllib.parse import urlparse


class DocumentPreprocessor:
    """
    Production-oriented document preprocessing for RAG pipelines.

    Supported Sources:
    - PDF text extraction output
    - Word documents
    - Web pages
    - Chat history
    - Emails
    - Markdown files
    - CSV/Text exports
    - OCR output
    """

    def __init__(self,
                 lowercase: bool = True,
                 remove_urls: bool = False,
                 remove_emails: bool = False,
                 remove_numbers: bool = False):
        self.lowercase = lowercase
        self.remove_urls = remove_urls
        self.remove_emails = remove_emails
        self.remove_numbers = remove_numbers

    def clean_document(self, text: str) -> Dict[str, Any]:
        """
        Clean a single document.
        """

        if text is None:
            return {
                "success": False,
                "error": "Document is None."
            }

        if not isinstance(text, str):
            return {
                "success": False,
                "error": f"Expected string, got {type(text).__name__}"
            }

        original_length = len(text)

        # --------------------------------------------------
        # HTML decoding
        # --------------------------------------------------
        text = html.unescape(text)

        # --------------------------------------------------
        # Remove HTML tags
        # --------------------------------------------------
        text = re.sub(r"<[^>]+>", " ", text)

        # --------------------------------------------------
        # Normalise unicode quotes/dashes
        # --------------------------------------------------
        text = text.replace(""", '"')
        text = text.replace(""", '"')
        text = text.replace(""", '"')
        text = text.replace(""", '"')
        text = text.replace("–", "-")
        text = text.replace("—", "-")

        # --------------------------------------------------
        # Remove page numbers
        # --------------------------------------------------
        text = re.sub(
            r"\bpage\s+\d+(\s+of\s+\d+)?\b",
            " ",
            text,
            flags=re.IGNORECASE
        )

        # --------------------------------------------------
        # Remove URLs
        # --------------------------------------------------
        if self.remove_urls:
            text = re.sub(r'http\S+|www\.\S+', ' ', text)

        # --------------------------------------------------
        # Remove emails
        # --------------------------------------------------
        if self.remove_emails:
            text = re.sub(
                r'\S+@\S+\.\S+',
                ' ',
                text
            )

        # --------------------------------------------------
        # Remove markdown syntax
        # --------------------------------------------------
        text = re.sub(r'#+\s*', '', text)
        text = re.sub(r'\*\*(.*?)\*\*', r'\1', text)
        text = re.sub(r'__(.*?)__', r'\1', text)
        text = re.sub(r'`(.*?)`', r'\1', text)

        # --------------------------------------------------
        # Remove repeated punctuation
        # --------------------------------------------------
        text = re.sub(r'([!?.,])\1+', r'\1', text)

        # --------------------------------------------------
        # Remove excessive symbols
        # --------------------------------------------------
        text = re.sub(
            r'[^\w\s.,!?;:()/-]',
            ' ',
            text
        )

        # --------------------------------------------------
        # Remove numbers if desired
        # --------------------------------------------------
        if self.remove_numbers:
            text = re.sub(r'\d+', ' ', text)

        # --------------------------------------------------
        # Case normalisation
        # --------------------------------------------------
        if self.lowercase:
            text = text.lower()

        # --------------------------------------------------
        # Remove excessive whitespace
        # --------------------------------------------------
        text = text.strip()
        text = text.lstrip()
        text = text.rstrip()

        text = text.replace("\n", " ")
        text = text.replace("\r", " ")
        text = text.replace("\t", " ")

        text = " ".join(text.split())

        # --------------------------------------------------
        # Sentence cleanup
        # --------------------------------------------------
        text = re.sub(r'\s+([.,!?])', r'\1', text)

        return {
            "success": True,
            "original_length": original_length,
            "cleaned_length": len(text),
            "cleaned_text": text
        }

    def process_documents(
        self,
        documents: List[str]
    ) -> Dict[str, Any]:
        """
        Process multiple documents.
        """

        if documents is None:
            return {
                "success": False,
                "error": "Documents cannot be None."
            }

        if not isinstance(documents, list):
            return {
                "success": False,
                "error": "Input must be a list."
            }

        processed = []
        failed = 0

        for doc in documents:
            result = self.clean_document(doc)

            if result["success"]:
                processed.append(result)
            else:
                failed += 1

        return {
            "success": True,
            "total_documents": len(documents),
            "processed_documents": len(processed),
            "failed_documents": failed,
            "documents": processed
        }


# ------------------------------------------------------
# Example data from various sources
# ------------------------------------------------------

sample_documents = [

    # PDF
    """
    Page 3 of 15

    Introduction to RAG Architecture

    Confidential Company Report
    """,

    # Website
    """
    <html>
        <body>
            Visit https://example.com for more information.
            Contact support@example.com
        </body>
    </html>
    """,

    # Chat History
    """
    John: Hey!!!
    Mary:   Can you send me the docs????
    John: Sure :)
    """,

    # Markdown
    """
    ## Embeddings

    **Embeddings** convert text into vectors.

    `python code`
    """,

    # OCR
    """
    TH1S  1S     OCR     OUTP UT

    PAGE     22
    """
]

processor = DocumentPreprocessor(
    remove_urls=True,
    remove_emails=True
)

results = processor.process_documents(
    sample_documents
)

import json
print(json.dumps(results, indent=2))
```

**Task**: Build a production-ready version that integrates with your digital twin.

```python
# # YOUR CODE HERE
# # Build your complete implementation

# class ProductionSystem:
#     """
#     Production-ready implementation for digital twin.
#     """

#     def __init__(self):
#         # Initialize your system
#         pass

#     def process(self, input_data):
#         # Add your implementation
#         pass

# Test your production system
# Add comprehensive tests here
from typing import List, Dict, Any
import logging
import re
import html
import json
import unicodedata
from datetime import datetime
from collections import Counter


class ProductionSystem:
    """
    Production-ready RAG preprocessing system.

    Features:
    ---------
    ✅ Multi-source document ingestion
    ✅ Data cleaning & normalization
    ✅ Comprehensive validation
    ✅ Error handling
    ✅ Logging
    ✅ Debug mode
    ✅ Analytics
    ✅ Visualization
    ✅ Metadata tracking
    """

    def __init__(self, debug: bool = False):

        self.debug = debug

        self.metrics = {
            "documents_processed": 0,
            "documents_failed": 0,
            "characters_processed": 0
        }

        self.processing_history = []

        logging.basicConfig(
            level=logging.DEBUG if debug else logging.INFO,
            format="%(asctime)s | %(levelname)s | %(message)s"
        )

        self.logger = logging.getLogger(__name__)

        self.logger.info("Production System Started")

    # ---------------------------------------------------
    # CLEANING PIPELINE
    # ---------------------------------------------------

    def clean_text(self, text: str) -> str:

        if not isinstance(text, str):
            raise TypeError(
                f"Expected string, got {type(text).__name__}"
            )

        # Unicode normalization
        text = unicodedata.normalize("NFKC", text)

        # Decode HTML entities
        text = html.unescape(text)

        # Remove HTML
        text = re.sub(r"<[^>]+>", " ", text)

        # Remove URLs
        text = re.sub(
            r"https?://\S+|www\.\S+",
            " ",
            text
        )

        # Remove emails
        text = re.sub(
            r"\S+@\S+\.\S+",
            " ",
            text
        )

        # Remove page markers
        text = re.sub(
            r"page\s+\d+",
            " ",
            text,
            flags=re.IGNORECASE
        )

        # Remove markdown
        text = re.sub(r'#+\s*', '', text)
        text = re.sub(r'\*\*(.*?)\*\*', r'\1', text)
        text = re.sub(r'`(.*?)`', r'\1', text)

        # Replace multiple whitespace
        text = re.sub(r"\s+", " ", text)

        # Remove repeated punctuation
        text = re.sub(
            r'([!?.,])\1+',
            r'\1',
            text
        )

        # Lowercase
        text = text.lower()

        # Trim
        text = text.strip()

        return text

    # ---------------------------------------------------
    # DOCUMENT PROCESSOR
    # ---------------------------------------------------

    def process(self, input_data: List[Dict[str, Any]]):

        if input_data is None:
            raise ValueError(
                "input_data cannot be None"
            )

        if not isinstance(input_data, list):
            raise TypeError(
                "input_data must be a list"
            )

        results = []

        for document in input_data:

            try:

                source_type = document.get(
                    "source",
                    "unknown"
                )

                content = document.get(
                    "content",
                    ""
                )

                cleaned = self.clean_text(
                    content
                )

                doc_result = {

                    "source": source_type,
                    "timestamp":
                        datetime.utcnow().isoformat(),

                    "original_length":
                        len(content),

                    "cleaned_length":
                        len(cleaned),

                    "cleaned_text":
                        cleaned
                }

                self.metrics[
                    "documents_processed"
                ] += 1

                self.metrics[
                    "characters_processed"
                ] += len(cleaned)

                results.append(
                    doc_result
                )

                self.logger.info(
                    f"Processed {source_type}"
                )

            except Exception as e:

                self.metrics[
                    "documents_failed"
                ] += 1

                self.logger.error(
                    str(e)
                )

                results.append({
                    "error": str(e)
                })

        self.processing_history.extend(
            results
        )

        return results

    # ---------------------------------------------------
    # ANALYTICS
    # ---------------------------------------------------

    def analytics(self):

        sources = []

        for item in self.processing_history:

            if "source" in item:
                sources.append(
                    item["source"]
                )

        source_counts = Counter(
            sources
        )

        return {
            "metrics": self.metrics,
            "source_distribution":
                dict(source_counts)
        }

    # ---------------------------------------------------
    # VISUALIZATION
    # ---------------------------------------------------

    def visualize_results(self):

        analytics = self.analytics()

        print("\nDOCUMENT SOURCES\n")

        for source, count in analytics[
            "source_distribution"
        ].items():

            print(
                f"{source:<15} | {'#' * count} ({count})"
            )

        print("\nSYSTEM METRICS\n")

        for key, value in analytics[
            "metrics"
        ].items():

            print(
                f"{key}: {value}"
            )

    # ---------------------------------------------------
    # DOCUMENTATION
    # ---------------------------------------------------

    def documentation(self):

        return """
        ProductionSystem Documentation
        ==================================

        Purpose:
        --------
        Production-ready preprocessing layer
        for Retrieval-Augmented Generation.

        Supported Sources:
        -------------------
        - PDF
        - Web Pages
        - Chat History
        - Emails
        - Markdown
        - OCR Output
        - JSON Records
        - Text Files

        Features:
        ---------
        1. Validation
        2. Cleaning
        3. Logging
        4. Analytics
        5. Visualization
        6. Metadata Tracking

        Pipeline:
        ---------
        Source
           ↓
        Validation
           ↓
        Cleaning
           ↓
        Metadata Creation
           ↓
        Analytics
           ↓
        Embedding Generation
           ↓
        Vector Store
        """


# ===================================================
# COMPREHENSIVE TESTS
# ===================================================

documents = [

    {
        "source": "pdf",
        "content":
        """
        Page 4
        Introduction to Retrieval Augmented Generation
        """
    },

    {
        "source": "web",
        "content":
        """
        <html>
            Visit https://example.com
            Contact: support@example.com
        </html>
        """
    },

    {
        "source": "chat",
        "content":
        """
        User: Hey!!!!!
        Bot: How can I help????
        """
    },

    {
        "source": "markdown",
        "content":
        """
        ## Embeddings

        **Embeddings are vector representations**
        """
    },

    {
        "source": "ocr",
        "content":
        """
        TH1S 1S OCR T3XT
        """
    }
]

# ===================================================
# RUN SYSTEM
# ===================================================

system = ProductionSystem(
    debug=True
)

results = system.process(
    documents
)

print("\nPROCESSED RESULTS")
print(
    json.dumps(
        results,
        indent=2
    )
)

print("\nANALYTICS")
print(
    json.dumps(
        system.analytics(),
        indent=2
    )
)

system.visualize_results()

print(
    system.documentation()
)
```
